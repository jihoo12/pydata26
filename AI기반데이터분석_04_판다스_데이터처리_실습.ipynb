## 1. 파이썬 파일 처리

```python
sample_file = 'data/sample.txt'
```

```python
# 파일을 open하여 전체 내용을 읽어서 출력
with open(sample_file,mode='rt') as f:
    contents = f.read()
    print(contents)
print("----- 끝 -----")
```

```python
# 파일을 open하고 한 줄씩 읽어서 출력하기
with open(sample_file,mode='r') as f:
    lines = f.readlines()
    for line in lines:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 파일객체에서 한 줄씩 읽어 오기
# 파일 객체 f를 반복문에서 사용하면 파일을 한 줄씩 읽으며,
# 각 줄 끝의 \n도 포함해서 가져옵
with open(sample_file, mode="r") as f:
    for line in f:
        print(line, end="")

print("----- 끝 -----")
```

```python
# 여러 개의 파일 읽기 (모든 txt 파일을 읽어서 출력하기)
import glob
files = glob.glob("data/*.txt")
print(files)

for file in files:
    with open(file,mode="r") as f:
        print(f.read())
    print("---- 끝 ----")
print("----- 끝 -----")
```

```python
# 지역별 판매 기록의 개수
sales_df['지역'].value_counts()
```

```python
# 지역과 상품의 조합별 판매 기록 개수
sales_df.value_counts(['지역','상품'])
```

```python
sample_df = pd.read_csv('data/sample_data.csv')
sample_df
```

```python
sample_df['City'].value_counts(dropna=False, ascending=True)
```

```python
# 지역별 총매출
sales_df.groupby(by='지역').sum()
```

```python
# 상품별 평균 매출
sales_df.groupby(by='상품')['매출'].mean()
```

```python
# 지역별 판매수량의 합계와 매출의 합계
sales_df.groupby(by='지역')[['판매수량','매출']].sum()
```

```python
# 지역과 상품별 평균 매출
sales_df.groupby(by=['지역','상품'])['매출'].mean()
```

```python
sales_df.pivot_table(
    values='매출',
    aggfunc='sum',
    index='지역',
    columns='상품'
)
```

```python
# 지역별·상품별 총매출을 구하고 행과 열의 합계 표시
sales_df.pivot_table(
    index='상품',
    columns='지역',
    values='매출',
    aggfunc='sum'
)
```

```python
# 지역별 매출 : 건수, 합계, 평균, 최대값
sales_df.groupby(by='지역')['매출'].agg(['count','sum','mean','max'])
```
