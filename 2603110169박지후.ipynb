```python
# import문 추가
import numpy as np

arr = np.arange(12).reshape(3,4)
print(arr)
```

```python
# 배열 구성 데이터 타입
print(arr.dtype)
```

```python
# 배열의 행과 열의 요소 갯수
print(arr.shape)
```

```python
# 배열 요소의 전체 갯수
print(arr.size)
```

```python
# 리스트로부터 넘파이 배열 생성
mylist = [[1, 2, 3], [4, 5, 6]]
a1 = np.array(mylist)

print(a1)
```

```python
# mylist로부터 실수형 넘파이 배열 생성
a2 = np.array(mylist,float)

print(a2)
```

```python
# 3 ~ 8 숫자로 1차원 배열 생성
a3 = np.arange(3,8)

print(a3)
```

```python
# 2 ~ 15 중 짝수로 1차원 배열 생성
a4 = np.arange(2,15,2)

print(a4)
```
