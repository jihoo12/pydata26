```python
numbers = [1, 2, 3, 4, 5]
# 시리즈 생성
data = pd.Series(number)
```

```python
# dictionary로부터 시리즈 생성하기
mart_dict = {'사과': 3000, '바나나': 1500, '수박': 12000, '포도': 5000}
data2 = pd.Series(mart_dict)

print(f"데이터 타입 (dtype): {data2.dtype}")
print(f"행과 열의 수 (shape): {data2.shape}")
print(f"데이터 수 (size): {data2.size}")

print(f"시리즈 이름 (name): {data2.name}") 

print(f"인덱스 (index): {data2.index}")
print(f"데이터 값 (values): {data2.values}")
```

### 3. 데이터프레임 객체 생성, 주요 속성 확인

```python
numbers = [1, 2, 3, 4, 5]
# 리스트로부터 데이터프레임 생성
num_df = pd.DataFrame(numbers)

# 컬럼명 'number'를 지정하여 데이터프레임 생성
num_df = pd.DataFrame(numbers,columns=['numbers'])
num_df
```

```python
score=[['정약용', 70, 80, 100],
       ['이순신', 100, 80, 70],
       ['이율곡', 90, 95, 90]]

# 2차원 리스트로 부터 데이터프레이 생성

score_df = pd.DataFrame(score)

score_df
```

```python
# 데이터프레임에 컬럼명 추가하기
names = ['이름', '국어', '영어', '수학']
score_df.columns = names
score_df
```

```python
# 데이터프레임 속성 확인 : 행과열의 수, 인덱스, 컬럼명, 데이터타입, 데이터값

print(score_df.shape,score_df.index,score_df.columns)
print(score_df.dtypes)
print(score_df.values)
```

```python
# 데이터프레임에 문자형 인덱스 추가하기
index = ['a', 'b', 'c']
score_df.index = index
score_df
```

```python
# dictionary로부터 데이터프레임 생성하기

score_dict={'이름': ['정약용', '이순신', '이율곡'],
            '국어' : [70, 100, 90],
            '영어' : [80, 80, 95],
            '수학' : [100, 70, 90]}

score_df = pd.DataFrame(score_dict)
score_df
```

```python
# '이름' 컬럼명을 'name'으로 변경하기

new_df = score_df.rename(columns={'이름':'name'})
new_df
```

```python
# 컬럼명을 모두 바꾸기 (영어로)
score_df.columns = ['name','korean','english','math']
score_df
```
