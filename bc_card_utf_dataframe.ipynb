```python
import pandas as pd
import re

file_path = "data/bc_card_utf.txt"

# 1. 파일 읽기
with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()

# 2. 헤더와 데이터 분리
# 원본 파일에서 마지막 컬럼명 CNT와
# 첫 번째 데이터의 REG_YYMM(201906)이 붙어 있음
header_text, data_text = text.split("CNT201906", 1) #CNT201906를 기준으로 한번 나누기

# 3. 컬럼명 복원
header_text = header_text.rstrip("\t") #오른쪽끝 tab제거
columns = header_text.split("\t") + ["CNT"] #tab기준으로 나누어 리스트로 만들고 "CNT"를 마지막 요소로 추가

# 4. 첫 번째 행의 REG_YYMM 복원
data_text = "201906" + data_text #data_text의 앞에 201906붙이기

# 5. 행 구분 복원
# 각 데이터 행이 201906으로 시작하므로
# 201906 앞에 줄바꿈 추가
data_text = re.sub(
    r"(?<!^)201906",
    "\n201906",
    data_text
) #201906찾아서 줄바꿈 넣기

# 6. 행 단위로 분리
lines = data_text.splitlines() #줄바꿈 기준으로 나눠서 리스트 만들기

# 7. 탭(\t)을 기준으로 각 행 분리
rows = [] #그냥 빈 리스트

for line in lines:
    values = line.split("\t") #tab으로 나누기

    # 정상적인 23개 컬럼을 가진 행만 저장
    if len(values) == len(columns): 
        rows.append(values) #values 길이랑 columns 길이랑 같으면 rows에 values 추가 

# 8. DataFrame 생성
df = pd.DataFrame(rows, columns=columns) 

# 9. 숫자형 컬럼 변환
numeric_cols = [
    "REG_YYMM",
    "MEGA_CTY_NO",
    "CTY_RGN_NO",
    "ADMI_CTY_NO",
    "MAIN_BUZ_CODE",
    "TP_GRP_NO",
    "TP_BUZ_NO",
    "CSTMR_MEGA_CTY_NO",
    "CSTMR_CTY_RGN_NO",
    "SEX_CTGO_CD",
    "FLC",
    "AMT",
    "CNT"
] 

df[numeric_cols] = df[numeric_cols].apply(
    pd.to_numeric,
    errors="coerce"
) #문자열로 저장되어있는 숫자 숫자 자료형으로 변환

# 결과 확인
print(df.shape) #출력
print(df.head()) #출력

df.info() #정보
```
